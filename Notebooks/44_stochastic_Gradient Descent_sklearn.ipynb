### `Stochastic Gradient Descent using Scikit-learn`

##### `Step 1 : Importing the required libraries`

```python
from sklearn.datasets import load_diabetes
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

##### `Step 2 :Splitting the dataset`

```python
X,y = load_diabetes(return_X_y=True)
print(X.shape)
print(y.shape)
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

##### `Step 3 : Fitting the linear Regression model on Training data`

```python
reg = LinearRegression()
reg.fit(X_train,y_train)

print(reg.coef_)
print(reg.intercept_)
```

##### `Step 4 : Evaluating the Model`

```python
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

##### `Step 5: Implementing SGD`

```python
from sklearn.linear_model import SGDRegressor
reg = SGDRegressor(max_iter=100,learning_rate='constant',eta0=0.01)
reg.fit(X_train,y_train)


y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
print("Weights (coefficients):", reg.coef_)
```

```python
print("Intercept:", reg.intercept_)
```

```python
# Make predictions
y_pred = reg.predict(X_test)
print("First 5 predictions:", y_pred[:5])
print("Actual values:", y_test[:5])
```

```python
# Compare with LinearRegression
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train, y_train)

print("LinearRegression coefficients:", lr.coef_)
print("SGDRegressor coefficients:", reg.coef_)
```

```python
print("Number of epochs run:", reg.n_iter_)
```

```python
# Increase max_iter and change learning rate
reg = SGDRegressor(max_iter=5000, learning_rate='invscaling', verbose=1)
reg.fit(X_train, y_train)
```

```python
# Scale the target variable
from sklearn.preprocessing import StandardScaler

y_scaler = StandardScaler()
y_train_scaled = y_scaler.fit_transform(y_train.reshape(-1,1)).ravel()

# Train the model with scaled target
reg.fit(X_train, y_train_scaled)
# Make predictions
y_pred_scaled = reg.predict(X_test)

# Inverse transform the predictions to original scale
y_pred = y_scaler.inverse_transform(y_pred_scaled.reshape(-1,1)).ravel()
```
