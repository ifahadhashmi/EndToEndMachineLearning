# `Ridge Regression`

```python
# import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# import dataset
from sklearn.datasets import load_diabetes

data=load_diabetes()
```

```python
# print dataset description
print(data.DESCR)
```

```python
# split dataset into features and target variable
X=data.data
y=data.target
```

```python
# split dataset into training and testing sets
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=45)
```

```python
# import and create Linear Regression model
from sklearn.linear_model import LinearRegression
L=LinearRegression()
```

```python
# fit the model
L.fit(X_train,y_train)
```

```python
# print coefficients and intercept
print(L.coef_)
print(L.intercept_)
```

```python
# make predictions
y_pred=L.predict(X_test)
```

```python
# evaluate the model, r2 score and RMSE
from sklearn.metrics import r2_score,mean_squared_error

print("R2 score",r2_score(y_test,y_pred))
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
# import and create Ridge Regression model
from sklearn.linear_model import Ridge
R=Ridge(alpha=100000)
```

```python
# fit the model
R.fit(X_train,y_train)
```

```python
# print coefficients and intercept
print(R.coef_)
print(R.intercept_)
```

```python
# make predictions
y_pred1=R.predict(X_test)
```

```python
# evaluate the model, r2 score and RMSE
print("R2 score",r2_score(y_test,y_pred1))
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred1)))
```

```python
# Generate and plot synthetic non-linear (polynomial) data with noise
m = 100
x1 = 5 * np.random.rand(m, 1) - 2
x2 = 0.7 * x1 ** 2 - 2 * x1 + 3 + np.random.randn(m, 1)

plt.scatter(x1, x2)
plt.show()
```

```python
# Fit Ridge Regression models with polynomial features of degree 16 for different alpha values
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures

def get_preds_ridge(x1, x2, alpha):
    model = Pipeline([
        ('poly_feats', PolynomialFeatures(degree=16)),
        ('ridge', Ridge(alpha=alpha))
    ])
    model.fit(x1, x2)
    return model.predict(x1)

alphas = [0, 20, 200]
cs = ['r', 'g', 'b']

plt.figure(figsize=(10, 6))
plt.plot(x1, x2, 'b+', label='Datapoints')

for alpha, c in zip(alphas, cs):
    preds = get_preds_ridge(x1, x2, alpha)
    # Plot
    plt.plot(sorted(x1[:, 0]), preds[np.argsort(x1[:, 0])], c, label='Alpha: {}'.format(alpha))

plt.legend()
plt.show()
```

```python
# Cross-validate Ridge Regression model on the diabetes dataset
from sklearn.linear_model import Ridge
from sklearn.model_selection import cross_val_score

ridge = Ridge(alpha=10)

scores = cross_val_score(
    ridge,
    X,
    y,
    cv=5,
    scoring='r2'
)

print(scores)
print("Average R2:", scores.mean())
```

```python
# import libraries
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV
```

```python
# parameter grid for alpha values
param_grid = {
    'alpha': [0.01, 0.1, 1, 10, 100]
}
```

```python
# create Ridge regression model
ridge = Ridge()

gs = GridSearchCV(
    estimator=ridge,
    param_grid=param_grid,
    cv=5,
    scoring='r2'
)
```

```python
# fit the model using grid search
gs.fit(X, y)
```

```python
# print best alpha and best score
print("Best alpha:", gs.best_params_)
print("Best score:", gs.best_score_)
```
